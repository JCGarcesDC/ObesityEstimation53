Esta sección sube automáticamente los modelos entrenados a Databricks usando las credenciales configuradas en `config/credentials.yaml`.

**Funcionalidades:**
- ✅ Conexión automática a Databricks
- 📊 Registro de métricas de todos los modelos
- 🏆 Registro del mejor modelo en MLflow Model Registry  
- 📁 Subida de artefactos adicionales

**Requisito:** `pip install mlflow` (se instalará automáticamente si no está disponible)

# **5.- Integración con Databricks**

```python
# Integración con Databricks - Subir modelos
import sys
import os
from pathlib import Path

# Configurar path para imports
src_path = Path('../../src').resolve()
sys.path.insert(0, str(src_path))

try:
    from utils.databricks_utils import setup_full_databricks_environment
    
    # Verificar que tenemos MLflow instalado
    try:
        import mlflow
        import mlflow.sklearn
    except ImportError:
        print("⚠️ MLflow no está instalado. Ejecutando: pip install mlflow")
        import subprocess
        subprocess.check_call([sys.executable, "-m", "pip", "install", "mlflow"])
        import mlflow
        import mlflow.sklearn
    
    # Configurar entorno Databricks
    setup_full_databricks_environment()
    
    # Configurar MLflow para Databricks
    mlflow.set_tracking_uri("databricks")
    
    # Usar el experimento existente en Databricks con la ruta completa
    experiment_name = "/Shared/ObesityEstimation"  # Ruta completa del experimento existente
    
    # Verificar conectividad
    print(f"MLflow URI: {mlflow.get_tracking_uri()}")
    
    # Verificar que el experimento existe y conectarse
    try:
        experiment = mlflow.get_experiment_by_name(experiment_name)
        if experiment is not None:
            print(f"📊 Conectando al experimento existente: {experiment_name}")
            mlflow.set_experiment(experiment_name)
        else:
            print(f"⚠️ No se encontró el experimento '{experiment_name}'. Disponibles:")
            experiments = mlflow.search_experiments()
            for exp in experiments[:5]:  # Mostrar solo los primeros 5
                print(f"  - {exp.name}")
            raise Exception(f"Experimento '{experiment_name}' no encontrado")
    except Exception as exp_error:
        print(f"❌ Error accediendo al experimento: {exp_error}")
        raise
    
    # Subir modelos a Databricks
    with mlflow.start_run(run_name="ObesityMine53_Model_Training"):
        
        # Log métricas de los mejores modelos (usando nombres correctos de columnas)
        for idx, row in df_resultados_opt.iterrows():
            modelo_nombre = row['Modelo']
            mlflow.log_metric(f"{modelo_nombre}_accuracy_test", row['Accuracy (Test)'])
            mlflow.log_metric(f"{modelo_nombre}_accuracy_train", row['Accuracy (Train)'])
            mlflow.log_metric(f"{modelo_nombre}_f1_test", row['F1-Score (Test)'])
            mlflow.log_metric(f"{modelo_nombre}_f1_train", row['F1-Score (Train)'])
        
        # Registrar el mejor modelo (basado en F1-Score Test)
        mejor_modelo_idx = df_resultados_opt['F1-Score (Test)'].idxmax()
        mejor_modelo_nombre = df_resultados_opt.loc[mejor_modelo_idx, 'Modelo']
        mejor_pipeline = pipelines_optimizados[mejor_modelo_nombre]
        
        # Log del mejor modelo (sin registered_model_name para evitar el error de Unity Catalog)
        try:
            mlflow.sklearn.log_model(
                mejor_pipeline,
                f"obesity_best_model_{mejor_modelo_nombre.replace(' ', '_')}",
                # registered_model_name comentado para evitar error de Unity Catalog
                # registered_model_name=f"ObesityMine53_BestModel_{mejor_modelo_nombre.replace(' ', '_')}"
            )
            print(f"✅ Modelo {mejor_modelo_nombre} loggeado exitosamente")
        except Exception as model_error:
            print(f"⚠️ Advertencia al registrar modelo: {model_error}")
            # Continuar sin registrar el modelo en el registry
        
        # Log resultados como artefacto
        results_path = "../../artefactos/mejores_modelos.csv"
        if os.path.exists(results_path):
            mlflow.log_artifact(results_path)
            print("✅ Artefactos subidos exitosamente")
        
        print(f"✅ Proceso completado exitosamente")
        print(f"🏆 Mejor modelo: {mejor_modelo_nombre}")
        print(f"📊 F1-Score (Test): {df_resultados_opt.loc[mejor_modelo_idx, 'F1-Score (Test)']:.4f}")
        print(f"📊 Experimento: {experiment_name}")
        print(f"🔗 URL: {os.environ.get('DATABRICKS_HOST')}")
        
except ImportError as e:
    print(f"⚠️ Error de importación: {e}")
    print("💡 Verifica que el path esté correcto y las utilidades estén disponibles")
except Exception as e:
    print(f"❌ Error en integración Databricks: {e}")
    print("🔧 Verifica la configuración de credenciales y conectividad")
```
